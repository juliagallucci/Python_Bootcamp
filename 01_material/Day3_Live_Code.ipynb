```python
#union combines two sets = all unique values in both

rainbow = {"red","orange","yellow","green","blue","indigo","violet"}
olympic_flag = {"red","green","yellow","blue","black"}
rainbow.union(olympic_flag)
```

```python
#intersection retains elements common to both
rainbow.intersection(olympic_flag)
```

```python
#symmetric diff; find values in either or but not both
rainbow.symmetric_difference(olympic_flag)
```

```python
#difference finds values unique only to the first set

rainbow.difference(olympic_flag)
```

```python
olympic_flag.difference(rainbow)
```
