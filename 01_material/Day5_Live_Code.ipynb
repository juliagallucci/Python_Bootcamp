```python
import matplotlib.pyplot as plt
```

```python
airlines = pd.read_csv('data/airlines.csv')
airlines
```

```python
plt.scatter(airlines["Delays_Weather"],
         airlines["Flights_Delayed"],
         marker = 's',
         facecolor = '#fb1',
         edgecolor = 'k')
plt.scatter(airlines['Delays_Security'],
            airlines['Flights_Delayed'],
            marker = 'o',
            facecolor = '#abc',
            edgecolors='k')
plt.show()
```

```python
fig, ax = plt.subplots()

ax.scatter(airlines['Delays_Weather'],
           airlines['Flights_Delayed'],
           marker = 's',
           facecolor = '#fb1',
           edgecolors= 'k', label = "Weather Delays")
ax.scatter(airlines['Delays_Security'],
           airlines['Flights_Delayed'],
           marker ='o',
           facecolor = '#abc',
           edgecolors='k', label = 'Security Delays')
ax.set_title("Specific Type of Delays vs Overall Flight Delays")
ax.set_xlabel("Specific Delays")
ax.set_ylabel("Overall Delays")
ax.legend()
plt.show()
```

```python
plt.style.available[5:10]
```

```python
plt.style.use('ggplot')
fig, ax = plt.subplots()

ax.scatter(airlines['Delays_Weather'],
           airlines['Flights_Delayed'],
           marker = 's',
           facecolor = '#fb1',
           edgecolors= 'k', label = "Weather Delays")
ax.scatter(airlines['Delays_Security'],
           airlines['Flights_Delayed'],
           marker ='o',
           facecolor = '#abc',
           edgecolors='k', label = 'Security Delays')
ax.set_title("Specific Type of Delays vs Overall Flight Delays")
ax.set_xlabel("Specific Delays")
ax.set_ylabel("Overall Delays")
ax.legend()
plt.show()
```

```python
small_airlines = airlines.head(n=10)
```

```python
bar_fig, bar_ax = plt.subplots()

bar_ax.barh(small_airlines['Airport_Code'],
            small_airlines['Late_Aircraft'])
bar_ax.set_title("Number of late Aircrafts for first 10 rows")
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(nrows = 2, ncols = 1)

ax1.barh(small_airlines['Airport_Code'],
         small_airlines['Late_Aircraft'])

ax2.scatter(small_airlines['Delays_Security'],
            small_airlines['Flights_Delayed'])

ax1.set_title("Number of late Aircrafts for each airport")
ax2.set_title("Security vs Overall Flight Delays")
plt.show()
```
