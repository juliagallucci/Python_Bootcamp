#### 2. Write a program to accept two integer numbers from the user and calculate their product (multiplication).

```python
def product():
  price = int(input("what is the price of your item"))
  quantity = int(input("how many do you need"))
  total = price * quantity
  print(f"your final price is {total}")
```
